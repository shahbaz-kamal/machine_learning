```python
import pandas as pd 

df1 =pd.DataFrame({
    "FellowshipId":[1001,1002,1003,1004],
    'FirstName':["Frodo","Samwise","Gandalf","Pippin"],
    'Skills':['Hiding','Garendening','Spells','Fireworks']
})

df2=pd.DataFrame({
     "FellowshipId":[1001,1002,1006,1007,1008],
     'FirstName':["Frodo","Samwise","Legolas","Elrond","Barromir"],
     'Age':[50,39,2931,6520,51]
})
display(df1)
display(df2)
```

```python
# inner join

mergedDf=df1.merge(df2)
mergedDf
mergedDf=df1.merge(df2,on=['FellowshipId','FirstName'],how='inner')
mergedDf
```

```python
# outer join

mergedDf=df1.merge(df2,how='outer')
mergedDf
```

```python
# left join

mergedDf=df1.merge(df2,how='left')
mergedDf
```

```python
# right join

mergedDf=df1.merge(df2,how='right')
mergedDf
```
