### Gradient descent calcualtion for simple linear regression

## initial value
- w=0
- alpha=0.01
- n=4
- b=0

```python
import pandas as pd

df=pd.DataFrame(
    {
        'x':[1,2,3,4],
        'y':[3,5,7,9]
    }
)
n=len(df)
alpha=0.01
w=1.758803967561962
b=1.722770737809606
gradients_with_weight=2/n * (1*(w*1-3+b)+ 2*(w*2-5+b)+3*(w*3-7+b)+4*(w*4-9+b))
gradients_with_bias= 2/n * ((w*1-3+b)+(w*2-5+b)+(w*3-7+b)+(w*4-9+b))

w_new= w -alpha*gradients_with_weight
b_new= w -alpha*gradients_with_bias
print("Data Points",n)
print("w",w)
print("b",b)
print("gradients_with_weight",gradients_with_weight)
print("gradients_with_bias",gradients_with_bias)
print("New weight",w_new)
print("New bias",b_new)
```
